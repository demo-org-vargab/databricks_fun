```python
dbutils.widgets.text("catalog_name",'')
dbutils.widgets.text("schema_name",'')
dbutils.widgets.text("admin_group",'')
dbutils.widgets.text("spn_id",'')
dbutils.widgets.text("dev_group",'')
dbutils.widgets.text("RO_group",'')
dbutils.widgets.text("RO_group2",'')
dbutils.widgets.text("logs_path",'')
```

```python
catalog_name = dbutils.widgets.get("catalog_name")
schema_name = dbutils.widgets.get("schema_name")
admin_group = dbutils.widgets.get("admin_group")
spn_id = dbutils.widgets.get("spn_id")
dev_group = dbutils.widgets.get("dev_group")
RO_group = dbutils.widgets.get("RO_group")
RO_group2 = dbutils.widgets.get("RO_group2")
logs_path = dbutils.widgets.get("logs_path")
```

```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()

w.schemas.create(
    catalog_name=catalog_name,
    name=schema_name,
    # storage_root = f"abfss://main@{catalog_name}.
    comment=f"Created by Databricks CLI for schema: {schema_name}"
)

print(f"Schema {schema_name} created in catalog {catalog_name}")
```

```python
w.schemas.update(
    full_name=f"{catalog_name}.{schema_name}",
    owner=f"{admin_group}"
)

print(f"set {admin_group} as owner in schema {catalog_name}.{schema_name}")
```

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import catalog

w = WorkspaceClient()

# Grant ALL PRIVILEGES on schema
w.grants.update(
    securable_type=catalog.SecurableType.SCHEMA,
    full_name=f"{catalog_name}.{schema_name}",
    changes=[
        catalog.PermissionsChange(
            principal=f"{spn_id}",
            add=[catalog.Privilege.ALL_PRIVILEGES]
        )
    ]
)

print(f"All permissions granted to {spn_id} on schema {catalog_name}.{schema_name}")
```

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import catalog

w = WorkspaceClient()

# Grant read-only permissions on schema
w.grants.update(
    securable_type=catalog.SecurableType.SCHEMA,
    full_name=f"{catalog_name}.{schema_name}",
    changes=[
        catalog.PermissionsChange(
            principal=f"{RO_group}",
            add=[
                catalog.Privilege.USE_SCHEMA,
                catalog.Privilege.SELECT,
                catalog.Privilege.READ_VOLUME,
                catalog.Privilege.EXECUTE
            ]
        )
    ]
)

print(f"Read-only permissions granted to {RO_group} on schema {catalog_name}.{schema_name}")
```
